# Outliers
# 27.07.2026

```python
df.columns
```

```python
print("Dataset Shape:",df.shape)
print("\n Coding_Score statistics before outlier treatment:")
print (df["Coding_Score"].describe())
```

```python
q1 = df["Study_Hours"].quantile(0.25)
q3 = df["Study_Hours"].quantile(0.75)

IQR = q3 - q1

lower_limit = q1 - 1.5 * IQR
upper_limit = q3 + 1.5 * IQR

outliers = df[
    (df["Study_Hours"] < lower_limit) |
    (df["Study_Hours"] > upper_limit)
]

print("\nStudy_Hours Outliers:", len(outliers))
print(outliers[["Student_ID", "Study_Hours"]].head(20))
```

```python
Q1 = df["CGPA"].quantile(0.25)
Q3 = df["CGPA"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df["CGPA_Capped"] = df["CGPA"].clip(lower=lower, upper=upper)
```

```python
print("After Outliers")

print("Original Minimum CGPA:", df["CGPA"].min())
print("Original Maximum CGPA:", df["CGPA"].max())

print("\nCapped Minimum CGPA:", df["CGPA_Capped"].min())
print("Capped Maximum CGPA:", df["CGPA_Capped"].max())

print("\nOriginal Mean CGPA:", df["CGPA"].mean())
print("Capped Mean CGPA:", df["CGPA_Capped"].mean())

print("\nOriginal Median CGPA:", df["CGPA"].median())
print("Capped Median CGPA:", df["CGPA_Capped"].median())
```

```python
X = df["Expected_Package_LPA"]
Y = df["CGPA_Capped"]
m = ((X -X.mean()*(Y-Y.mean()))).sum() / ((X -X.mean())**2).sum()
c = Y.mean() -m* X.mean()
Y_pred = m * X + c
print(Y_pred.head())
#Y_pred = np.round(Y_pred,0)
Y_pred = Y_pred.round(0)
print(Y_pred)
Y_pred = np.round(Y_pred).astype(int)
print(Y_pred.head())
print("Slope = " ,m)
print("Intercept =",c)
```

```python
plt.figure(figsize=(8,5))
plt.scatter(X,Y)
plt.plot(X,Y_pred,color = "red")
plt.title("simple Linear chart")
plt.xlabel("Expected_Package_LPA")
plt.ylabel("CGPA_Capped")
plt.show()
```

```python
import matplotlib.pyplot as plt

sorted_data = pd.DataFrame({
    "X": X,
    "Actual": Y,
    "Predicted": Y_pred
}).sort_values("X")

plt.figure(figsize=(8,5))

plt.plot(sorted_data["X"], sorted_data["Actual"], label="CGPA Salary")
plt.plot(sorted_data["X"], sorted_data["Predicted"], label="CGPA Salary")

plt.title("Actual vs Predicted Salary")
plt.xlabel("Expected_Package_LPA")
plt.ylabel("CGPA_Capped")
plt.legend()
plt.grid(True)

plt.show()
```

# LINEAR REGRESSION
# 01.08.2026

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Load dataset
df = pd.read_csv(r"C:\Users\vaddi\OneDrive\Desktop\AIML\Preprocessed data.csv")

# Select required columns
df = df[['CGPA_Capped', 'Expected_Package_LPA']]

# Remove missing values
df = df.dropna()

# Define X and y
X = df[['CGPA_Capped']]
y = df['Expected_Package_LPA']

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Train the model
model = LinearRegression()
model.fit(X_train, y_train)

# Print model parameters
print("Intercept (b0):", model.intercept_)
print("Slope (b1):", model.coef_[0])

print("\nRegression Equation:")
print(f"Expected_Package_LPA = {model.intercept_:.2f} + ({model.coef_[0]:.2f} × CGPA_Capped)")

# Prediction
y_pred = model.predict(X_test)

# Result table
result = pd.DataFrame({
    "CGPA_Capped": X_test["CGPA_Capped"],
    "Actual Salary": y_test,
    "Predicted Salary": y_pred
})

print(result.head(10))

# Plot data
plot_data = pd.DataFrame({
    "CGPA_Capped": X_test["CGPA_Capped"],
    "Actual Salary": y_test,
    "Predicted Salary": y_pred
})

plot_data = plot_data.sort_values("CGPA_Capped")

plt.figure(figsize=(8,5))

plt.scatter(
    plot_data["CGPA_Capped"],
    plot_data["Actual Salary"],
    color="blue",
    label="Actual Salary"
)

plt.plot(
    plot_data["CGPA_Capped"],
    plot_data["Predicted Salary"],
    color="red",
    linewidth=2,
    label="Regression Line"
)

plt.title("Linear Regression using Scikit-learn")
plt.xlabel("CGPA_Capped")
plt.ylabel("Expected_Package_LPA")
plt.legend()
plt.grid(True)
plt.show()
```
