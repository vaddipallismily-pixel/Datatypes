```python
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    def display(self):
        print(f"Employee ID: {self.emp_id}")
        print(f"Employee Name: {self.name}")
        print(f"Salary: {self.salary}\n")

emp1 = Employee(101, "Rahul", 50000)
emp2 = Employee(102, "Anjali", 65000)


emp1.display()
emp2.display()
```

# Probability

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("student_performance_1000.csv")

# Students with attendance >= 90
high_attendance = df[df["Attendance"] >= 90]

# Students with marks > 80 among them
high_marks = high_attendance[high_attendance["Marks"] > 80]

# Probability
probability = len(high_marks) / len(high_attendance)

print("Probability of scoring >80 given Attendance ≥90:", probability)
```

# OUTLINE

```python
import matplotlib.pyplot as plt

# Boxplot before outlier treatment
plt.figure(figsize=(8,3))
plt.boxplot(df["price"])
plt.title("Price Before Outlier Treatment")
plt.ylabel("Price")
plt.show()
```

```python
# Calculate IQR
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
# Remove outliers
df = df[(df["price"] >= lower) & (df["price"] <= upper)]
```

```python
# Boxplot after outlier treatment
plt.figure(figsize=(8,3))
plt.boxplot(df["price"])
plt.title("Price After Outlier Treatment")
plt.ylabel("Price")
plt.show()
```
